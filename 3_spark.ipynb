##### Spark:
- Runs in memory (very fast)
- Works on clusters (many machines)
- Handles TBs–PBs of data
- Supports SQL, Python, Java, Scala, R

##### Core Spark Components

-  Spark Ecosystem
   - Spark Core – Basic execution engine
   - Spark SQL – Structured data (tables, SQL)
   - Spark Streaming – Real-time data
   - MLlib – Machine learning
   - GraphX – Graph processing

##### Spark Architecture (Very Important)
- Master–Worker Model

##### Key Terms
- Driver → Your main program
- Executor → Runs tasks
- Cluster Manager → Manages resources (YARN, Kubernetes, Standalone)
