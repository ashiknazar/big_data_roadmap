## Apache Spark Architecture (Simplified)

### 1. Cluster Manager
- Manages cluster resources such as CPU and memory.
- Examples: Standalone, YARN, Kubernetes, Mesos.
- Does **not** execute user code.

### 2. Driver Program
- Started when a Spark application is submitted.
- Runs the user’s main application code.
- Responsibilities:
  - Creates the `SparkContext`
  - Converts transformations into a **DAG (Directed Acyclic Graph)**
  - Splits the DAG into **jobs, stages, and tasks**
  - Requests resources from the Cluster Manager
  - Schedules tasks on executors

### 3. Executors
- Launched by the Cluster Manager on worker machines.
- JVM processes that:
  - Execute tasks sent by the Driver
  - Store data in memory or disk (cache)
  - Send results and status back to the Driver

### 4. Execution Flow
- User submits Spark application.
- Driver:
  - Builds DAG from transformations
  - Breaks DAG into **jobs → stages → tasks**
- Cluster Manager:
  - Allocates resources
  - Starts executors
- Executors:
  - Execute tasks in parallel
  - Return results to the Driver

### 5. Key Points
- Spark follows a **driver–executor architecture**, not master–slave.
- Driver coordinates execution; executors perform the work.
- Cluster Manager only manages resources.
