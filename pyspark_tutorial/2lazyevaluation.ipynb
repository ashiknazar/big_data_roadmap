## Lazy Evaluation in Apache Spark

- Spark uses **lazy evaluation**, which means:
  - Transformations are **not executed immediately** when they are defined.

### How It Works
- When you apply **transformations** (e.g., `map`, `filter`, `select`):
  - Spark only **records the operations**
  - It builds a **logical execution plan (DAG)**

- No data is processed at this stage.

### When Execution Happens
- Execution starts only when an **action** is called.
- Examples of actions:
  - `count()`
  - `collect()`
  - `show()`
  - `save()`

### What Happens on Action
- Spark:
  - Optimizes the DAG
  - Splits it into **jobs, stages, and tasks**
  - Executes the tasks on executors

### Why Spark Uses Lazy Evaluation
- Enables **query optimization**
- Avoids unnecessary computations
- Improves **performance and efficiency**
- Allows Spark to decide the best execution plan

### Simple Example
- `df.filter(...).select(...)` → No execution
- `df.filter(...).select(...).count()` → Execution starts

### Key Point
- **Transformations = Lazy**
- **Actions = Trigger execution**
