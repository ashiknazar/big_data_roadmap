## Spark Execution Concepts

### 🔹 Transformations
- Operations that **define how data should be processed**
- They are **lazy** → execution does not happen immediately
- Only create a **logical plan**
- Examples:
  - `map()`
  - `filter()`
  - `select()`
  - `groupBy()`

📌 Transformations build a **DAG** but do not execute it.

---

### 🔹 DAG (Directed Acyclic Graph)
- A **logical execution plan** created by Spark
- Nodes represent **RDDs / DataFrames**
- Edges represent **transformations**
- Called *acyclic* because it **does not loop back**

📌 Spark builds the DAG from transformations to understand **dependencies**.

---

### 🔹 Jobs
- A **job** is created when an **action** is called
- Each action triggers **one job**
- Examples of actions:
  - `count()`
  - `collect()`
  - `show()`
  - `save()`

📌 A job represents a complete computation requested by the user.

---

### 🔹 Stages
- A **job is divided into multiple stages**
- Each stage contains tasks that can run **in parallel**
- Stages are separated by **shuffle operations**
- Types:
  - **Shuffle Stage**
  - **Result Stage**

📌 A stage groups operations that do not require data shuffling.

---

### 🔹 Tasks
- The **smallest unit of execution**
- One task runs on **one partition** of data
- Tasks are executed on **worker nodes**
- Multiple tasks run in parallel across the cluster

📌 Number of tasks = number of partitions in a stage.
