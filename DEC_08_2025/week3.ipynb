##### File Permissions & User/Group Management

#### Topics
- Ownership: user, group, others
- chmod for permissions (r = 4, w = 2, x = 1)
- chown, chgrp basics
- Superuser privileges (sudo)
- Creating users & groups

| Purpose            | Commands                                        |
| ------------------ | ----------------------------------------------- |
| Check permissions  | `ls -l`                                         |
| Change permissions | `chmod 755 filename`                            |
| Change owner/group | `chown user:group filename`                     |
| Switch user        | `su - username`                                 |
| Add user/group     | `sudo adduser johndoe`, `sudo groupadd dataeng` |
