#### 1️⃣ HDFS (Concepts First, Commands Second)

Must-know concepts

- NameNode vs DataNode
- Blocks & replication
- Rack awareness (high-level)
- Read/write path

___

##### NameNode (Master)
- Stores metadata, not actual data
- Maintains:
  - File names
  - Directory structure
  - File permissions
  - Mapping of files → blocks → DataNodes
- Decides:
  - Where blocks are stored
  - How many replicas exist
- Handles:
  - Client requests (read/write)
  - DataNode health monitoring (via heartbeats)

- What it does NOT do:
  - ❌ Does not store file data
  - ❌ Does not handle data transfer

- Single point of control (but not failure if HA is enabled)
   - High Availability
   - Active NameNode,Standby NameNode

##### DataNode(Worker)

- Stores actual data blocks
- Reads/writes blocks as instructed by NameNode
- Sends 
  - Heartbeats (I’m alive)
  - Block reports (what data it has)
- Replicates data blocks to other DataNodes
