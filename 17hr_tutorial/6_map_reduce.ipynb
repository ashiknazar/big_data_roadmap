## Java vs Hadoop MapReduce Data Types

| Java Data Type | Hadoop MapReduce Writable Equivalent |
|----------------|--------------------------------------|
| `int`          | `IntWritable`                        |
| `long`         | `LongWritable`                       |
| `float`        | `FloatWritable`                      |
| `double`       | `DoubleWritable`                     |
| `boolean`      | `BooleanWritable`                    |
| `char`         | `Text` (no direct char, use String)  |
| `String`       | `Text`                               |
| `byte[]`       | `BytesWritable`                      |
| `Object`       | `Writable` (interface, custom types) |
| `Array/List`   | `ArrayWritable`                      |
| `Map`          | `MapWritable`                        |
| `Null`         | `NullWritable`                       |
<br>
<br>

> **Note:** All Hadoop Writable types implement the `Writable` and `WritableComparable` interfaces, enabling serialization and sorting in MapReduce.
