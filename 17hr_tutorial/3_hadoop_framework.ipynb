### <u>Pig</u>


- yahoo
- piglatin is the scripting language
 

- we can do **same operation** with pig and hive but sql is more familiar 

- Apache Pig is a high-level data flow scripting platform built on top of Hadoop.
- Pig Latin is the scripting language used in Apache Pig.
- Simplify ETL (Extract, Transform, Load) tasks and data analysis without writing raw MapReduce code.
- Think of Pig Latin as a "bridge" between SQL and MapReduce.
- Procedural Language → unlike SQL (declarative), Pig tells the system how to process data
- Handles Semi-structured data → JSON, log files, XML.
- Extensible → you can write UDFs (User Defined Functions) in Java, Python, or other languages.
- Execution → Pig Latin scripts get converted into MapReduce jobs (or Tez/Spark jobs).

| Feature              | **Hive** (HiveQL) 🟦                                      | **Pig** (Pig Latin) 🟨                                                |
| -------------------- | --------------------------------------------------------- | --------------------------------------------------------------------- |
| **Language Style**   | Declarative (SQL-like) → *what* to do                     | Procedural (scripting) → *how* to do                                  |
| **Target Users**     | SQL developers, data analysts                             | Programmers, data engineers                                           |
| **Use Cases**        | Querying structured data (tables, partitions)             | ETL (Extract, Transform, Load), working with raw/semi-structured data |
| **Data Types**       | Works best with structured/tabular data                   | Handles structured + semi/unstructured (logs, JSON, XML)              |
| **Execution Engine** | MapReduce / Tez / Spark                                   | MapReduce / Tez / Spark                                               |
| **Ease of Use**      | Easier for analysts (similar to SQL)                      | More flexible for complex data flows                                  |
| **Output**           | Often used for BI/Reports (connects to Tableau, Power BI) | Often used in data pipelines & transformations                        |


- **3. Export (Hadoop → RDBMS)**
- Exports processed data from HDFS back into MySQL.
```sqoop
sqoop export \
--connect jdbc:mysql://localhost:3306/retail \
--username root \
--password 1234 \
--table sales_summary \
--export-dir /user/hadoop/sales_summary
```

### <u>FLume</u>

- like sqoop a data pipeline component
- retrieve data in real time
- mqueue (messaging queue)
- sqoop is only rdbms and Flume is not only
- only incoming of data to hadoop is available for flume unlike sqoop


- Apache Flume is a **data ingestion** tool designed to collect, aggregate, and move large amounts of streaming data (usually log data) into Hadoop’s HDFS or HBase.
- In big data pipelines, we often need to move real-time or near real-time data (like logs, events, or sensor data) into Hadoop.
- Writing custom MapReduce/Java code for ingestion was hard → Flume solved that problem with a ready-made ingestion framework.
___

#### **Flume Architecture**
 - It works on a **Source → Channel → Sink model**:
   - Source: Collects data (e.g., log files, syslog, HTTP, Avro, Kafka).
 - Channel: A temporary store/buffer (memory or file) that holds data until it’s consumed.
 - Sink: Delivers data to its final destination (e.g., HDFS, HBase, Kafka).
