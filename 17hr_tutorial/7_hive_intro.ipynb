### hive

- after running hive server we can send request to it with CLI
- when it get a request to create a table , it already have a metadata (containing all hive table info - which is different from meta data of name node)
- hive will store this meta data information in rdbms only
- ![](hive.png)

- if hive rdbms is not available to store meta data ,hive will have an embedded metastore rdbms called derby.remote metastore
- in realtime they wont use it
-  ![](derby.png)

- if create table request is send to node 3 and is stored in derby which is local to that machine and another request to insert is send to node one , whose derby does not have info about the table
