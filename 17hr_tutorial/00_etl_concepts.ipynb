# ETL (Extract, Transform, Load) in Big Data

## What is ETL?

ETL stands for:

- **E**xtract
- **T**ransform
- **L**oad

It is the process of collecting data from different sources, cleaning and transforming it, and loading it into a target system such as a Data Warehouse, Data Lake, or Analytics Platform.

---

# ETL Architecture

```text
+----------------+
| Data Sources   |
+----------------+
        |
        v
+----------------+
|   Extract      |
+----------------+
        |
        v
+----------------+
|  Transform     |
+----------------+
        |
        v
+----------------+
|     Load       |
+----------------+
        |
        v
+----------------+
| Data Warehouse |
| Data Lake      |
+----------------+
```

---

# 1. Extract

## Purpose

Collect data from multiple sources.

### Common Data Sources

- Relational Databases (MySQL, PostgreSQL, Oracle)
- NoSQL Databases (MongoDB, Cassandra)
- CSV Files
- JSON Files
- XML Files
- APIs
- Log Files
- Social Media Data
- IoT Devices
- Streaming Data (Kafka)

### Example

Customer data stored in:

- MySQL database
- CSV sales files
- Website logs

ETL extracts all these datasets.

---

# 2. Transform

## Purpose

Convert raw data into a clean and useful format.

### Common Transformations

### Data Cleaning

Remove:

- Null values
- Duplicate records
- Invalid data

Example:

```text
Before:
John,25
John,25
NULL,30

After:
John,25
```

---

### Data Standardization

Convert data into a common format.

Example:

```text
01/01/2025
2025-01-01
Jan 1 2025

Converted To:

2025-01-01
```

---

### Data Filtering

Keep only useful records.

Example:

```sql
SELECT *
FROM sales
WHERE amount > 1000;
```

---

### Data Aggregation

Summarize data.

Example:

```text
Daily Sales:

100
200
300

Total Sales = 600
```

---

### Data Enrichment

Add information from another source.

Example:

```text
Customer ID → Customer Name
```

---

### Data Validation

Check:

- Data types
- Business rules
- Constraints

Example:

```text
Age cannot be negative.
```

---

# 3. Load

## Purpose

Store transformed data in the target system.

### Target Systems

- Data Warehouse
- Data Lake
- Hadoop HDFS
- Hive
- Spark Tables
- Cloud Storage
- SQL Databases

### Example

```text
Cleaned Customer Data
        |
        v
Data Warehouse
```

# ETL vs ELT

## ETL

```text
Extract
   |
Transform
   |
Load
```

Transformation occurs before loading.

---

## ELT

```text
Extract
   |
Load
   |
Transform
```

Transformation occurs after loading.

Used heavily in modern cloud platforms.

Examples:

- Snowflake
- BigQuery
- Databricks

# Interview Questions

### 1. What is ETL?

ETL is the process of Extracting data from sources, Transforming it into a useful format, and Loading it into a target system.

### 2. What is the difference between ETL and ELT?

ETL transforms data before loading, while ELT loads data first and transforms later.

### 3. Why is Spark preferred for ETL?

Because Spark performs in-memory distributed processing, making ETL much faster than traditional MapReduce.

### 4. Which Hadoop tools are commonly used for ETL?

- Sqoop
- Flume
- Hive
- Pig
- MapReduce
- Spark

### 5. What is real-time ETL?

Continuous processing of incoming data streams using tools such as Kafka and Spark Streaming.

# Slowly Changing Dimensions (SCD)

A Slowly Changing Dimension (SCD) is a dimension whose attributes change over time.

Examples:

- Customer changes address
- Employee changes department
- Product changes category
- Supplier changes location

The challenge is deciding how to handle these changes.

---

# SCD Type 0 – Fixed Dimension

No changes allowed.

Example:

| Customer ID | Name |
|------------|------|
| 101 | John |

Even if the source changes, the warehouse record remains unchanged.

Use Case:
- Date of Birth
- Original Registration Date

---

# SCD Type 1 – Overwrite

Update the existing record.

Before:

| Customer ID | City |
|------------|------|
| 101 | Mumbai |

After:

| Customer ID | City |
|------------|------|
| 101 | Bangalore |

History is NOT preserved.

Advantages:
- Simple
- Requires less storage

Disadvantages:
- Historical data lost

---

# SCD Type 2 – Preserve Full History

Insert a new record whenever a change occurs.

Before:

| SK | Customer ID | City | Current |
|----|------------|------|---------|
| 1 | 101 | Mumbai | Y |

After:

| SK | Customer ID | City | Current |
|----|------------|------|---------|
| 1 | 101 | Mumbai | N |
| 2 | 101 | Bangalore | Y |

History is preserved.

Additional Columns:

- Surrogate Key
- Start Date
- End Date
- Current Flag

Most commonly asked SCD type in interviews.

---

# SCD Type 3 – Limited History

Store previous value in an additional column.

Before:

| Customer ID | Current City | Previous City |
|------------|-------------|--------------|
| 101 | Mumbai | NULL |

After:

| Customer ID | Current City | Previous City |
|------------|-------------|--------------|
| 101 | Bangalore | Mumbai |

Only limited history is maintained.

---

# SCD Type 4 – History Table

Current data stored in one table.

Historical data stored in another table.

Example:

Customer_Current

| Customer ID | City |
|------------|------|
| 101 | Bangalore |

Customer_History

| Customer ID | City |
|------------|------|
| 101 | Mumbai |

---

# SCD Type 6 (Hybrid)

Combination of:

- Type 1
- Type 2
- Type 3

Provides:
- Current value
- Previous value
- Complete history

Used in advanced data warehouse designs.

---

# Comparison

| Type | History Preserved | Method |
|--------|----------------|---------|
| Type 0 | No Changes Allowed | Ignore Updates |
| Type 1 | No | Overwrite |
| Type 2 | Yes | New Row |
| Type 3 | Limited | New Column |
| Type 4 | Yes | History Table |
| Type 6 | Yes | Hybrid Approach |
